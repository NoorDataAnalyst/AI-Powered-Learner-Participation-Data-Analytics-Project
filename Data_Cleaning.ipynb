FORMATTING INCONSISTENTS

```python
text_columns = ['Gender', 'Country', 'Opportunity Category', 'Status Description', 'Institution Name']
for col in text_columns:
  print(f"\n Unique values check for: {col}")
  print(df[col].value_counts(dropna=False).head(10))
```

```python
# Convert date columns
date_columns = [
    'Learner SignUp DateTime', 'Date of Birth', 'Entry created at',
    'Apply Date', 'Opportunity Start Date', 'Opportunity End Date'
]
for col in date_columns:
    df[col] = pd.to_datetime(df[col], errors='coerce')

#  Log timestamp
log_time = pd.Timestamp.now().strftime('%Y-%m-%d %H:%M:%S')

#  Identify logic errors
error_end_before_start = df['Opportunity End Date'] < df['Opportunity Start Date']
error_apply_after_end = df['Apply Date'] > df['Opportunity End Date']

print(f"[{log_time}] End Date < Start Date: {error_end_before_start.sum()} rows")
print(f"[{log_time}] Apply Date > End Date: {error_apply_after_end.sum()} rows")

# Fix swapped Start/End dates instead of deleting the rows
swap_mask = error_end_before_start
old_start = df.loc[swap_mask, 'Opportunity Start Date'].copy()
df.loc[swap_mask, 'Opportunity Start Date'] = df.loc[swap_mask, 'Opportunity End Date']
df.loc[swap_mask, 'Opportunity End Date'] = old_start
print(f"[{log_time}] Swapped Start/End dates for: {swap_mask.sum()} rows")

# Re-check for any remaining logic errors after the swap
still_bad = (df['Opportunity End Date'] < df['Opportunity Start Date']) | (df['Apply Date'] > df['Opportunity End Date'])
print(f"[{log_time}] Remaining logic errors after swap: {still_bad.sum()} rows")

# Only drop rows if still invalid after the swap attempt
df = df[~still_bad].copy()

print(f"[{log_time}] Final dataset size: {len(df)} rows")
```

```python

# Force text type, remove spaces, standardize capitalization
df['First Name'] = df['First Name'].astype(str).str.strip().str.lower().str.title()

#  Replace names with numbers or symbols as 'Unknown'
df['First Name'] = df['First Name'].replace(r'.*[\d@?#!$%^&*()].*', 'Unknown', regex=True)

#Handle missing text variations
df['First Name'] = df['First Name'].replace(['Nan', 'None', ''], np.nan)

# Quick check: how many were converted to Unknown?
unknown_count = (df['First Name'] == 'Unknown').sum()
print(f"Rows converted to 'Unknown': {unknown_count}")

# Quick check: sample of cleaned names
print("\n of clean first names:")
print(df['First Name'].dropna().head(10))
```

```python
#Clean: remove extra spaces and standardize capitalization for ALL institutions
df['Institution Name'] = df['Institution Name'].str.strip().str.title()

#Define the cleaning for Saint Louis University variations
standard_name_cleaning = {
    'Saint Louis University': 'Saint Louis University',
    'Saint Louis Univeristy': 'Saint Louis University',
    'St Louis University': 'Saint Louis University',
    'St. Louis University': 'Saint Louis University',
    'Slu': 'Saint Louis University',
    'Sl U': 'Saint Louis University',
    'Saint Louis Univ': 'Saint Louis University',
    'Saint Louis': 'Saint Louis University',
    'Saint Louis University ': 'Saint Louis University',
    'Saint Louise University': 'Saint Louis University',
    'Saint Louis Univerisity': 'Saint Louis University',
    'Saint Louis Univesity': 'Saint Louis University',
    'Saint Louis Universy': 'Saint Louis University',
    'Saint Louis University-Main Campus': 'Saint Louis University',
    'Saint Louis University I': 'Saint Louis University',
    'Saint Louis University  Saint Louis Missouri': 'Saint Louis University',
    'Saint Louis Unveracity': 'Saint Louis University',
    'Saint Louis Universty': 'Saint Louis University',
    'Saint Louis Univercity': 'Saint Louis University',
    'Saint Louis University  Mo': 'Saint Louis University',
    'Saint Louis  University': 'Saint Louis University',
    'Saint Louis Unniversity': 'Saint Louis University',
    'Saint Louis University- Madrid': 'Saint Louis University',
    'Saint Louisuniversity': 'Saint Louis University',
    'University Of Missouri Saint Louis': 'Saint Louis University',
    'Saint Louis University Madrid': 'Saint Louis University',
    'Saint Louis University-Madrid': 'Saint Louis University',
}

#Apply the cleaning
df['Institution Name'] = df['Institution Name'].replace(standard_name_cleaning)

#Fill missing values
df['Institution Name'] = df['Institution Name'].fillna('Unknown')

# Verify the cleaning
print(df['Institution Name'].value_counts().head(10))
```
